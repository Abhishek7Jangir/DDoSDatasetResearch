```python
dfs = {"df1": c17, "df2": c18, "df3": c19, "df4": booter,
       "df5": caida, "df6": botiot, "df7": ctu}

# 1. Number of columns in each
for name, df in dfs.items():
    print(name, "->", df.shape[1], "columns")

# 2. Columns common to ALL dataframes
common = set.intersection(*(set(df.columns) for df in dfs.values()))
print("\nCommon columns:", len(common))

# 3. Columns that are NOT in all dataframes
all_cols = set.union(*(set(df.columns) for df in dfs.values()))
different = all_cols - common
print("Columns not shared by all:", len(different))

# 4. For each differing column, which dataframes have it
for col in sorted(different):
    present_in = [name for name, df in dfs.items() if col in df.columns]
    print(f"  {col!r}: present in {present_in}")
```

```python
import numpy as np
import pandas as pd

label_col = "Label"
df = combined

num_cols = df.select_dtypes(include=np.number).columns

# Row-level flags
dup_mask = df.duplicated()                                  # extra copies of a row (first one not counted)
nan_mask = df.isna().any(axis=1)                            # rows with at least one NaN
inf_mask = np.isinf(df[num_cols]).any(axis=1)               # rows with at least one +inf/-inf

# ---------- Overall ----------
print("Total rows:", len(df))
print("\n1. Duplicate rows:", dup_mask.sum())
print("2. Rows with NaN:", nan_mask.sum(), "| total NaN cells:", df.isna().sum().sum())
print("3. Rows with inf:", inf_mask.sum(), "| total inf cells:", np.isinf(df[num_cols]).sum().sum())

# ---------- Per class ----------
summary = pd.DataFrame({
    "total_rows":     df.groupby(label_col).size(),
    "duplicates":     dup_mask.groupby(df[label_col]).sum(),   # 4
    "rows_with_nan":  nan_mask.groupby(df[label_col]).sum(),   # 5
    "rows_with_inf":  inf_mask.groupby(df[label_col]).sum(),   # 6
})
summary["dup_%"] = (summary["duplicates"] / summary["total_rows"] * 100).round(2)
summary["nan_%"] = (summary["rows_with_nan"] / summary["total_rows"] * 100).round(2)
summary["inf_%"] = (summary["rows_with_inf"] / summary["total_rows"] * 100).round(2)

print("\nPer-class breakdown:")
print(summary)

# ---------- Which columns have the NaN / inf ----------
col_report = pd.DataFrame({
    "nan_count": df.isna().sum(),
    "inf_count": np.isinf(df[num_cols]).sum(),
}).fillna(0).astype(int)
print("\nColumns with NaN or inf:")
print(col_report[(col_report["nan_count"] > 0) | (col_report["inf_count"] > 0)])
```

```python
before = len(combined)

combined.drop_duplicates(inplace=True)
combined.dropna(inplace=True)
combined.reset_index(drop=True, inplace=True)

print("Rows before:", before)
print("Rows after:", len(combined))
print("Removed:", before - len(combined))
```
