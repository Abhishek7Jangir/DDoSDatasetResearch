```python
df_attack = df_attack.rename(columns={
    "Fwd Pkts/s": "Fwd Packets/s",
    "Bwd Pkts/s": "Bwd Packets/s",
    "Bwd IAT Tot": "Bwd IAT Total"
})
```

```python
# Column names
benign_cols = list(df_benign.columns)
attack_cols = list(df_attack.columns)

print("Same columns:", benign_cols == attack_cols)

# If not the same, show differences
print("Only in benign:", set(benign_cols) - set(attack_cols))
print("Only in attack:", set(attack_cols) - set(benign_cols))
```

```python
df_attack = df_attack[df_benign.columns]
```

```python
df_attack = df_attack.astype(df_benign.dtypes.to_dict())
```

```python
df_benign["Label"] = df_benign["Label"].astype(int)
df_attack["Label"] = df_attack["Label"].astype(int)
```

```python
assert list(df_benign.columns) == list(df_attack.columns)
assert (df_benign.dtypes == df_attack.dtypes).all()
```

```python
import pandas as pd

df_combined = pd.concat(
    [df_benign, df_attack],
    ignore_index=True
)
```

```python
print(df_combined.shape)
print(df_combined["Label"].value_counts())
print(df_combined.dtypes.value_counts())
```

```python
df_combined.to_csv(
    "CaidaDDoSWithBenignCombined.csv",
    index=False,
    chunksize=100_000
)
```
