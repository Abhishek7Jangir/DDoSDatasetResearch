```python
import pandas as pd

# File paths
small_file = "caida_dataset_2007_Combined.csv"
big_file = "CombinedBoTIoT_&_CIC2019.csv"

# Step 1: Load small file into memory
df_small = pd.read_csv(small_file)
```

```python
df_small['Label'].value_counts()
```

```python
# Step 2: Rename columns to match big file
df_small = df_small.rename(columns={
    'Fwd Pkts/s': 'Fwd Packets/s',
    'Bwd Pkts/s': 'Bwd Packets/s',
    'Bwd IAT Tot': 'Bwd IAT Total'
})

# Step 3: Get big file column order
big_columns = pd.read_csv(big_file, nrows=0).columns.tolist()

# Step 4: Reorder small file columns to match big file
df_small = df_small[big_columns]

# Step 5: Fix int->float mismatches
float_fix_cols = ['Bwd IAT Total', 'Bwd IAT Mean', 'Bwd IAT Std', 'Bwd IAT Max', 'Bwd IAT Min']
for col in float_fix_cols:
    if df_small[col].dtype != 'float64':
        df_small[col] = df_small[col].astype('float64')

# Step 6: Verify columns match exactly
columns_match = df_small.columns.tolist() == big_columns
print("Columns match with big file:", columns_match)
if not columns_match:
    raise ValueError("Column names/order do not match the big file. Aborting append.")

# Step 7: Append directly to big CSV
# mode='a' -> append, header=False -> skip writing header
df_small.to_csv(big_file, mode='a', index=False, header=False)

print("Small file appended to big file successfully!")
```
