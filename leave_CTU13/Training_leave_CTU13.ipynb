```python
# --- 2. REPRODUCIBILITY (The "Same Result" Rule) ---
def set_seeds(seed=42):
    os.environ['PYTHONHASHSEED'] = str(seed)
    os.environ['TF_DETERMINISTIC_OPS'] = '1'
    os.environ['TF_CUDNN_DETERMINISTIC'] = '1'
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)
    tf.keras.utils.set_random_seed(seed)
    print(f"Environment seeds locked at {seed}")

set_seeds(42) 
```

```python
# Load the specific binary dataset
df = pd.read_csv("CombinedBoTIoT_&_CIC2019.csv")

# Verify the load by checking the total rows and columns
print(f"Dataset successfully loaded.")
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
```

```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
# 1. Strip any leading/trailing whitespace and convert to lowercase
df.columns = df.columns.str.strip().str.lower()

# 2. Replace spaces, slashes, and dots with underscores for easy coding
df.columns = df.columns.str.replace(' ', '_', regex=False)
df.columns = df.columns.str.replace('/', '_', regex=False)
df.columns = df.columns.str.replace('.', '_', regex=False)

# 3. Print the first 10 columns to verify the change
print("First 10 standardized columns:")
print(df.columns[:10].tolist())

# 4. Find the exact name of your Label column
label_col = [col for col in df.columns if 'label' in col]
print(f"\nLabel column found: {label_col}")
```

```python
# List of columns that are not useful for training features
identifiers = ['flow_id', 'src_ip', 'dst_ip', 'timestamp']

# Drop them only if they exist in the current dataframe
df.drop(columns=[col for col in identifiers if col in df.columns], inplace=True)

print(f"Identifiers removed.")
print(f"Remaining columns: {df.shape[1]}")
```

```python
rows_with_nan = df.isna().any(axis=1).sum()

print(f"Total rows with at least one NaN: {rows_with_nan}")
```

```python
# ==========================================================
import numpy as np
import gc
from sklearn.linear_model import LinearRegression

print("Starting Chunked Regression Imputation (Memory Safe)...")

# 1. Replace all Inf with NaN as per the paper's methodology
df.replace([np.inf, -np.inf], np.nan, inplace=True)

# 2. Identify columns with and without NaN
all_features = df.drop(columns=['label']).columns
COLS_WNA = df.columns[df.isna().any()].tolist()
COLS_WONA = [c for c in all_features if c not in COLS_WNA]

# We will save models here immediately to avoid running a second loop later
imputer_models = {} 

# 3. Memory-Safe Regression Imputation Loop
chunk_size = 5_000_000  # Safe chunk size for 128GB RAM

for target_col in COLS_WNA:
    print(f"Processing column: {target_col}")

    # Use indices instead of making massive copies of the DataFrame
    train_idx = df.index[df[target_col].notna()]
    predict_idx = df.index[df[target_col].isna()]
    
    if len(predict_idx) > 0:
        n_features = len(COLS_WONA)
        XTX = np.zeros((n_features + 1, n_features + 1), dtype=np.float64)
        XTy = np.zeros(n_features + 1, dtype=np.float64)
        
        # --- TRAINING PHASE (Chunked) ---
        for i in range(0, len(train_idx), chunk_size):
            idx_chunk = train_idx[i : i + chunk_size]
            X_chunk = df.loc[idx_chunk, COLS_WONA].values
            y_chunk = df.loc[idx_chunk, target_col].values
            X_chunk_int = np.hstack([X_chunk, np.ones((len(X_chunk), 1), dtype=np.float64)])
            
            XTX += np.dot(X_chunk_int.T, X_chunk_int)
            XTy += np.dot(X_chunk_int.T, y_chunk)
            del X_chunk, y_chunk, X_chunk_int

        # Solve for weights
        try:
            beta = np.linalg.solve(XTX, XTy)
        except np.linalg.LinAlgError:
            beta = np.dot(np.linalg.pinv(XTX), XTy)

        # Inject weights into standard sklearn model
        model = LinearRegression()
        model.coef_ = beta[:-1]      
        model.intercept_ = beta[-1]  
        model.n_features_in_ = n_features
        model.feature_names_in_ = np.array(COLS_WONA)
        
        imputer_models[target_col] = model

        # --- PREDICTION PHASE (Chunked) ---
        for i in range(0, len(predict_idx), chunk_size):
            idx_chunk = predict_idx[i : i + chunk_size]
            X_pred_chunk = df.loc[idx_chunk, COLS_WONA].values
            preds = model.predict(X_pred_chunk)
            df.loc[idx_chunk, target_col] = preds
            del X_pred_chunk, preds
            
    gc.collect() # Force clear RAM

print(f"Imputation Complete. Total NaN remaining: {df.isna().sum().sum()}")
# ==========================================================
```

```python
import joblib

# Simply dump the dictionary we populated in the previous step
joblib.dump({
    'models': imputer_models,
    'cols_wona': COLS_WONA,
    'cols_wna': COLS_WNA
}, "regression_imputer.pkl")

print("Regression Imputer models saved to regression_imputer.pkl")
```
