```python
import pandas as pd

reference_file = "CTU-13.csv"
output_file = "leave_AnonBooter.csv"

# Read only headers from reference file
df_ref = pd.read_csv(reference_file, nrows=0)

df_ref = df_ref.rename(columns={
    'Fwd Pkts/s': 'Fwd Packets/s',
    'Bwd Pkts/s': 'Bwd Packets/s',
    'Bwd IAT Tot': 'Bwd IAT Total'
})

# Create empty CSV with same columns
df_ref.to_csv(output_file, index=False)

print(f"Created empty CSV: {output_file} with columns from {reference_file}")
```

```python
import os
import pandas as pd

small_file = "CTU-13.csv"
big_file = "leave_AnonBooter.csv"
n = 5933  # number of records with Label=1
m = 0   # number of records with Label=0

# --- Step 1: Load small file ---
df_small = pd.read_csv(small_file)
```

```python
df_small['Label'].value_counts()
```

```python
# # Convert label to binary
# df_small['Label'] = (df_small['Label'] != 'Benign').astype(int)
```

```python
# Step 2: Rename columns to match big file
df_small = df_small.rename(columns={
    'Fwd Pkts/s': 'Fwd Packets/s',
    'Bwd Pkts/s': 'Bwd Packets/s',
    'Bwd IAT Tot': 'Bwd IAT Total'
})

# Step 3: Get column order from big file (IMPORTANT STEP YOU USED BEFORE)
big_columns = pd.read_csv(big_file, nrows=0).columns.tolist()

# Step 4: Select n and m rows
df_label_1 = df_small[df_small['Label'] == 1].head(n)
df_label_0 = df_small[df_small['Label'] == 0].head(m)

# Combine
df_to_append = pd.concat([df_label_1, df_label_0], ignore_index=True)

# Step 5: Ensure exact column order match with big file
df_to_append = df_to_append[big_columns]

# Step 6: Append to big file
df_to_append.to_csv(
    big_file,
    mode='a',
    index=False,
    header=False
)

print(f"Appended {len(df_to_append)} rows (Label=1: {len(df_label_1)}, Label=0: {len(df_label_0)})")
```
