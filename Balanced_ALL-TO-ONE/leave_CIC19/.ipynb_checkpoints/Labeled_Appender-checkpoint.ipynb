```python
import pandas as pd

# 1. Target your master file directly
target_file = "appended_labeledFIle.csv"

# 2. Put the exact labels you want to erase here
labels_to_erase = [
    "benign_cicids2017", 
    "attack_cicids2017"
]

print(f"Loading {target_file} into memory...")
df = pd.read_csv(target_file)
original_count = len(df)
print(f"Original row count: {original_count}")

# 3. Filter out the mistake labels
df_cleaned = df[~df['Label'].isin(labels_to_erase)]

new_count = len(df_cleaned)
rows_removed = original_count - new_count

print(f"\nRemoved {rows_removed} rows containing the mistaken labels.")
print(f"New row count: {new_count}")

# 4. Overwrite the exact same file
print(f"\nOverwriting {target_file} with cleaned data...")
df_cleaned.to_csv(target_file, index=False)

print("SUCCESS! The mistaken records are gone. You can now re-run append script for 10,000.")
```

```python
import pandas as pd
import numpy as np

# ==========================================
# 1. CONFIGURATION (CHANGE THESE FOR EACH RUN)
# ==========================================
current_file_path = "anon-Booter_Combined.csv"   # The file you want to process right now
current_dataset_name = "booters"    # The tag to append (e.g., cic17, cic18)

# Global settings
big_file = "appended_labeledFIle.csv"                  
n = 10000                        # Target number of Benign & Attack rows

print(f"Starting processing for dataset: {current_dataset_name.upper()}")
print("-" * 50)

# ==========================================
# 2. LOAD, ALIGN & CLEAN DATA
# ==========================================
big_columns = pd.read_csv(big_file, nrows=0).columns.tolist()

df = pd.read_csv(current_file_path)

df = df.rename(columns={
    'Fwd Pkts/s': 'Fwd Packets/s',
    'Bwd Pkts/s': 'Bwd Packets/s',
    'Bwd IAT Tot': 'Bwd IAT Total'
})

# Clean pure NaNs and Infinities in the features
print(f"Original row count: {len(df)}")
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)

# ==========================================
# 3. THE BULLETPROOF LABEL ROUTER
# ==========================================
# Convert everything to a string, remove spaces, and force uppercase
df['Label'] = df['Label'].astype(str).str.strip().str.upper()

# Master routing lists
known_benign_indicators = ["BENIGN", "0", "NORMAL"]

# We only trash completely broken label text now. 
# "NO LABEL" is omitted from here, meaning it will safely pass through to the attacks.
known_invalid_indicators = ["NONE", "NULL", "NAN", ""]

# 1. The Trash Bin: Drop legitimately broken labels
rows_before_trash = len(df)
df = df[~df['Label'].isin(known_invalid_indicators)]
dropped_invalid = rows_before_trash - len(df)
if dropped_invalid > 0:
    print(f"Cleaned Label count: {len(df)} (Dropped {dropped_invalid} corrupted labels)")

# 2. Safely split the remaining pure data
# "NO LABEL" is not Benign, so it gets perfectly routed into df_attack_source here
df_benign_source = df[df['Label'].isin(known_benign_indicators)]
df_attack_source = df[~df['Label'].isin(known_benign_indicators)]

# ==========================================
# 4. PROCESS BENIGN RECORDS
# ==========================================
if not df_benign_source.empty:
    benign_sample_size = min(n, len(df_benign_source))
    df_benign = df_benign_source.sample(n=benign_sample_size, random_state=42).copy()
    df_benign['Label'] = f"benign_{current_dataset_name}"
else:
    df_benign = pd.DataFrame()
    benign_sample_size = 0

# ==========================================
# 5. AUTO-DETECT & PROCESS ATTACK RECORDS
# ==========================================
if not df_attack_source.empty:
    attack_classes = df_attack_source['Label'].unique()
    num_classes = len(attack_classes)
    target_per_class = n // num_classes
    
    print(f"Auto-detected {num_classes} distinct legitimate attack class(es).")
    print(f"Targeting {target_per_class} records per class.")
    
    attack_samples = []
    
    for attack_class in attack_classes:
        class_data = df_attack_source[df_attack_source['Label'] == attack_class]
        sample_size = min(target_per_class, len(class_data))
        attack_samples.append(class_data.sample(n=sample_size, random_state=42).copy())
        
    df_attack = pd.concat(attack_samples, ignore_index=True)
    df_attack['Label'] = f"attack_{current_dataset_name}"
else:
    df_attack = pd.DataFrame()

# ==========================================
# 6. COMBINE, ALIGN, AND APPEND
# ==========================================
df_to_append = pd.concat([df_benign, df_attack], ignore_index=True)

if not df_to_append.empty:
    # Safely align columns (fills missing ones with NaN)
    df_to_append = df_to_append.reindex(columns=big_columns)
    
    # Append to the big file
    df_to_append.to_csv(big_file, mode='a', index=False, header=False)
    
    print("\nSUCCESS! Summary of appended data:")
    print(f"  -> Benign Rows: {benign_sample_size} (Labeled as 'benign_{current_dataset_name}')")
    print(f"  -> Attack Rows: {len(df_attack)} (Labeled as 'attack_{current_dataset_name}')")
    print(f"  -> Total Appended: {len(df_to_append)} rows")
else:
    print("\nWARNING: No valid records found to append!")

print("-" * 50)
```
