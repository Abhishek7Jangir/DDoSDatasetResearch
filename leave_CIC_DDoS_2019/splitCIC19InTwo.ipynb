```python
input_file = "CIC-DDoS-2019_training.csv"

# -------- First pass: count total rows (excluding header) --------
with open(input_file, "rb") as f:
    total_rows = -1  # start at -1 to exclude header
    for _ in f:
        total_rows += 1

split_point = total_rows // 2  # half the rows

# -------- Second pass: split the file --------
with open(input_file, "rb") as infile, \
     open("part1CIC2019.csv", "wb") as f1, \
     open("part2CIC2019.csv", "wb") as f2:

    # Read and write header to both files
    header = infile.readline()
    f1.write(header)
    f2.write(header)

    for i, line in enumerate(infile):
        if i < split_point:
            f1.write(line)
        else:
            f2.write(line)
```
